```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(500)
```

```python
n_sample = 1000
class_0_ratio = 0.9

n_class_0 = int(n_sample * class_0_ratio)
n_class_1 = n_sample - n_class_0
```

```python
n_class_1,n_class_0
```

```python
class_0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})
```

```python
class_0
```

```python
class_1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
class_1
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
df
```

```python
df['target'].value_counts()
```

***We are doing Upscaling here:***

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample

df_minority_upsampled = resample(
    df_minority,
    replace=True,
    n_samples=len(df_majority),
    random_state=42
    )
```

```python
df_minority_upsampled.shape
```

```python
df_upsampled= pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```

***Upscaling Done.*** 💖

***Downsampling*** **-------------->**

```python
class_3 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_4 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

df_n = pd.concat([class_3,class_4]).reset_index(drop=True)
```

```python
df_maj_ority = df_n[df_n['target']==0]
df_min_ority = df_n[df_n['target']==1]
```

```python
from sklearn.utils import resample

df_maj_ority_upsampled = resample(
    df_maj_ority,
    replace=True,
    n_samples=len(df_min_ority),
    random_state=50
    )
```

```python
df_upsampled_2= pd.concat([df_maj_ority_upsampled, df_min_ority])

df_upsampled_2['target'].value_counts()
```
