## Positional Argument

```python
def print_details (*args):
    for item in args:
        print(item)
```

```python
print_details(10,20,'Hello', 'Gello')
```

## Keyword Argument

```python
def print_details (**keyargs):
    for item in keyargs.items():
        print(item)
```

```python
print_details (age='six', man= 1)
```

```python
def print_details (**keyargs):
    for key, value in keyargs.items():
        print(key, value)
```

```python
print_details(name='Rezon', age= 27, wife='Rozy')
```

## Mix

```python
def print_details (*args, **keyargs):
    for item in args:
        print(item)
    for key, value in keyargs.items():
        print(f"{key} -> {value}")
```

```python
print_details(10, 20, 50, name='Rezon', age= 27, console='Pc')
```

```python
def print_details (args):
    for item in args:
        print(item)
```

```python
print_details([10,20,'Hello', 'Gello'])
```
