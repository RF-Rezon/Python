# Feature Engineering 

```python
df['Date'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head(2)
```

```python
df.info()
```

```python
df['Date'] = df['Date'].astype(dtype=int)
df['Month'] = df['Month'].astype(dtype=int)
df['Year'] = df['Year'].astype(dtype=int)
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis=1, inplace=True)
```

```python
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x: x.split(' ')[0])
df['Arrival_Hour'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_Minute'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.drop('Arrival_Time', axis=1, inplace=True)
```

```python
df.head(3)
```

```python
df['Arrival_Hour'] = df['Arrival_Hour'].astype(int)
df['Arrival_Minute'] = df['Arrival_Minute'].astype(int)
```

```python
df['Dep_Hour']=df['Dep_Time'].str.split(':').str[0]
df['Dep_Min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.head(3)
```

```python
df.drop('Dep_Time',axis=1,inplace=True)

df['Dep_Hour'] = df['Dep_Hour'].astype(int)
df['Dep_Min'] = df['Dep_Min'].astype(int)
```

```python
df.head(3)
```

```python
df['Arrival_Min'] = df['Arrival_Minute']
```

```python
df.head(2)
```

```python
df.drop('Arrival_Minute', axis=1, inplace=True)
```

# Handeling Categorical 

```python
df['Total_Stops'].unique()
```

```python
# df['Total_Stops']=df['Total_Stops'].map({'non-stop':0, '2 stops':2, '1 stop':1, '3 stops':3, np.nan:1, '4 stops':4})
```

```python
df['Total_Stops'] = df['Total_Stops'].fillna('1 stop')
```

```python
df['Total_Stops'].unique()
```

```python
encoder = OrdinalEncoder(
    categories=[['non-stop','1 stop','2 stops','3 stops','4 stops']]
) # choto theke boro ba kom gurutto theke beshi gurutto er dike. 

df[['Total_Stops']] = encoder.fit_transform(df[['Total_Stops']])
```

```python
df.head(3)
```

```python
df.drop('Route', axis=1, inplace=True)
```

```python
df.tail(3)
```

```python
df['Duration_Hour'] = df['Duration'].str.extract('(\d+)h') # h er age ja digit ase dhoro
df['Duration_Min']  = df['Duration'].str.extract('(\d+)m') # m er age ja digit ase dhoro

df['Duration_Hour'] = df['Duration_Hour'].fillna(0).astype(int)  
df['Duration_Min']  = df['Duration_Min'].fillna(0).astype(int)
```

```python
df.drop('Duration', axis=1, inplace=True)
```

```python
df.info()
```

```python
df.head(4)
```

```python
df['Additional_Info'].value_counts()
```

```python
df.drop(columns=['Additional_Info'], inplace=True)
```
