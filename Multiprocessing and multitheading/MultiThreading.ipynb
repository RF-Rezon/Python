## Multithreading

```python
import time
import threading

# Function 1
def task1():
    print("Task 1 started")
    time.sleep(3)
    print("Task 1 finished")


# Function 2
def task2():
    print("Task 2 started")
    time.sleep(3)
    print("Task 2 finished")


# ----------------------------
# Without Threading
# ----------------------------
start = time.time()

task1()
task2()

end = time.time()

print("\nWithout Threading Time:", end - start)


# ----------------------------
# With Threading
# ----------------------------
start = time.time()

t1 = threading.Thread(target=task1)
t2 = threading.Thread(target=task2)

t1.start()
t2.start()

t1.join()
t2.join()

end = time.time()

print("\nWith Threading Time:", end - start)
```
