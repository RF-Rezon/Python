```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('Algerian_forest_fires_dataset.csv', header=1)
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df = df.drop(index=122).reset_index(drop=True)
```

```python
df[df.isnull().any(axis=1)]
```

```python
df.columns = df.columns.str.strip()
df.columns
```

```python
df['Classes']
```

```python
df['Classes'].value_counts()
```

```python
df['Classes'] = df['Classes'].astype(str).str.strip().str.lower()
```

```python
df['Classes'].value_counts()
```

```python
df = df[df['Classes'].isin(['fire', 'not fire'])]
df = df.reset_index(drop=True)
```

```python
df['Classes'].value_counts()
```

```python
df
```

```python
df['Classes'] = (df['Classes'] == 'fire').astype(int)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.columns
```

```python
df[['Temperature', 'RH', 'Ws']] = df[['Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

```python
df[['Rain', 'FFMC','DMC', 'DC', 'ISI', 'BUI', 'FWI']] = df[['Rain', 'FFMC','DMC', 'DC', 'ISI', 'BUI', 'FWI']].astype(float)
```
