```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red','blue','green','purple','red','purple','blue','yellow']
})
```

```python
df.head()
```

```python
encoder = OneHotEncoder()
```

```python
sparse = encoder.fit_transform(df[['color']]).toarray()
# # or
# sparse = encoder.fit_transform(df).toarray()
```

```python
final_df = pd.DataFrame(data=sparse, columns=encoder.get_feature_names_out())
```
